```python
import random

def student_performance_analyzer():
    print("--- Student Performance Data Analyzer ---")
    
    last_name = input("Enter your LAST_NAME: ").strip().upper()
    favorite_artist = input("Enter your FAVORITE_ARTIST: ").strip().upper()
    try:
        seed_num = int(input("Enter your SEED_NUM (Integer): "))
    except ValueError:
        print("Invalid seed number. Defaulting to 42.")
        seed_num = 42

    random.seed(seed_num)

    # Core Configurations
    subjects = ['PROGRAMMING', 'MATHEMATICS', 'ELECTRONICS', 'PHYSICS', 'ENGLISH']
    students = [f"{last_name}_S{i}" for i in range(1, 6)]
    
    artist_offsets = [ord(char) for char in favorite_artist if char.isalnum()]
    if not artist_offsets:
        artist_offsets = [42]  # Fallback

    def get_performance_category(score):
        if score is None:
            return "INCOMPLETE"
        if score >= 90: return "EXCELLENT"
        elif score >= 80: return "VERY GOOD"
        elif score >= 70: return "GOOD"
        elif score >= 60: return "PASS"
        else: return "FAIL"

    raw_records = []
    offset_idx = 0
    
    for student in students:
        for subj in subjects:
            base_score = random.randint(50, 95)
            modifier = artist_offsets[offset_idx % len(artist_offsets)] % 6  
            score = min(100, base_score + modifier)
            
            raw_records.append({"student": student, "subject": subj, "score": score})
            offset_idx += 1

    if raw_records:
        repeated_target = raw_records[0]
        raw_records.append({
            "student": repeated_target["student"],
            "subject": repeated_target["subject"],
            "score": min(100, repeated_target["score"] + 2) 
        })

    raw_records.append({
        "student": students[-1], 
        "subject": subjects[-1], 
        "score": None # Missing score
    })

    seen_pairs = set()
    repeated_records = []
    incomplete_records = []
    distinct_records = []

    for rec in raw_records:
        rec["category"] = get_performance_category(rec["score"])
        
        if rec["score"] is None:
            incomplete_records.append(rec)
        
        pair = (rec["student"], rec["subject"])
        if pair in seen_pairs:
            repeated_records.append(rec)
        else:
            seen_pairs.add(pair)
            distinct_records.append(rec)

    student_summaries = {}
    for st in students:
        scores = [r["score"] for r in raw_records if r["student"] == st and r["score"] is not None]
        if scores:
            student_summaries[st] = {"avg": round(sum(scores)/len(scores), 2), "count": len(scores)}

    subject_summaries = {}
    for subj in subjects:
        scores = [r["score"] for r in raw_records if r["subject"] == subj and r["score"] is not None]
        if scores:
            subject_summaries[subj] = {"avg": round(sum(scores)/len(scores), 2), "count": len(scores)}

    ordered_results = sorted(raw_records, key=lambda x: (x["student"], x["subject"]))

    print("\nAssessment Data")
    
    print("\n1. Generated Performance Records (Raw Output):")
    for idx, r in enumerate(raw_records, 1):
        print(f"   [{idx:02d}] Student: {r['student']} | Subject: {r['subject']} | Score: {r['score']}")

    print("\n2. Organized Student/Subject Data Map:")
    organized_map = {}
    for r in raw_records:
        organized_map.setdefault(r["student"], []).append(f"{r['subject']}({r['score']})")
    for st, data in organized_map.items():
        print(f"   {st} -> {', '.join(data)}")

    print("\n3. Repeated / Incomplete Records Found:")
    print("   [Repeated rows]:")
    for r in repeated_records:
        print(f"   - Duplicate performance link found for {r['student']} in {r['subject']}")
    print("   [Incomplete rows]:")
    for r in incomplete_records:
        print(f"   - Missing/Unrecorded metric field detected for {r['student']} in {r['subject']}")

    print("\n4. Performance Categories Analysis:")
    for r in ordered_results:
        print(f"   {r['student']} | {r['subject']:<12} | Score: {str(r['score']):<4} | Category: {r['category']}")

    print("\n5. Student and Subject Performance Summaries:")
    print("   [By Student Averages]:")
    for st, metrics in student_summaries.items():
        print(f"   - {st:<12}: GPAR Average = {metrics['avg']} ({metrics['count']} valid tests)")
    print("   [By Subject Averages]:")
    for subj, metrics in subject_summaries.items():
        print(f"   - {subj:<12}: Course Average = {metrics['avg']}")

    print("\n6. Ordered Results Overview (Sorted Alphabetically):")
    print(f"   {'Student Name':<15} | {'Subject Course':<15} | {'Final Score':<11} | {'Status/Category'}")
    print("-" * 65)
    for r in ordered_results:
        print(f"   {r['student']:<15} | {r['subject']:<15} | {str(r['score']):<11} | {r['category']}")

    print("Final Output: Student Performance Report Completed Successfully.")

if __name__ == "__main__":
    student_performance_analyzer()
```
