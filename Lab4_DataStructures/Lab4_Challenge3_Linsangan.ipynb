```python
import random

def transaction_history_analyzer():
    print("--- Transaction History Analyzer ---")

    last_name = input("Enter your LAST_NAME: ").strip().upper()
    favorite_artist = input("Enter your FAVORITE_ARTIST: ").strip().upper()
    try:
        seed_num = int(input("Enter your SEED_NUM (Integer): "))
    except ValueError:
        print("Invalid seed number. Defaulting to 42.")
        seed_num = 42
    random.seed(seed_num)

    products = ['NOTEBOOK', 'PEN', 'USB', 'HEADPHONES', 'CALCULATOR']
    customers = [f"{last_name}_C{i}" for i in range(1, 6)]

    fixed_prices = {
        'NOTEBOOK': 45.00,
        'PEN': 12.50,
        'USB': 350.00,
        'HEADPHONES': 750.00,
        'CALCULATOR': 1200.00
    }

    artist_offsets = [ord(char) for char in favorite_artist if char.isalnum()]
    if not artist_offsets:
        artist_offsets = [7] 

    raw_transactions = []
    offset_idx = 0

    for cust in customers:

        sampled_products = random.sample(products, k=random.randint(3, 4))
        for prod in sampled_products:
            base_qty = random.randint(1, 5)
            artist_mod = (artist_offsets[offset_idx % len(artist_offsets)] % 3)
            quantity = base_qty + artist_mod
            unit_price = fixed_prices[prod]
            total_value = round(quantity * unit_price, 2)
            
            raw_transactions.append({
                "customer": cust,
                "product": prod,
                "quantity": quantity,
                "unit_price": unit_price,
                "total_value": total_value
            })
            offset_idx += 1
    if raw_transactions:
        repeated_target = raw_transactions[0]
        raw_transactions.append({
            "customer": repeated_target["customer"],
            "product": repeated_target["product"],
            "quantity": random.randint(1, 2), 
            "unit_price": repeated_target["unit_price"],
            "total_value": round(random.randint(1, 2) * repeated_target["unit_price"], 2)
        })

    seen_pairs = set()
    repeated_txs = []
    distinct_tx_count = 0

    for tx in raw_transactions:
        pair = (tx["customer"], tx["product"])
        if pair in seen_pairs:
            repeated_txs.append(tx)
        else:
            seen_pairs.add(pair)
            distinct_tx_count += 1

    customer_summaries = {}
    for cust in customers:
        cust_txs = [t for t in raw_transactions if t["customer"] == cust]
        if cust_txs:
            customer_summaries[cust] = {
                "total_spent": round(sum(t["total_value"] for t in cust_txs), 2),
                "total_items": sum(t["quantity"] for t in cust_txs)
            }

    product_summaries = {}
    for prod in products:
        prod_txs = [t for t in raw_transactions if t["product"] == prod]
        product_summaries[prod] = {
            "units_sold": sum(t["quantity"] for t in prod_txs),
            "revenue": round(sum(t["total_value"] for t in prod_txs), 2)
        }

    ordered_results = sorted(raw_transactions, key=lambda x: (x["customer"], x["product"]))

    print("\nAssessment Data")
    
    print("\n1. Generated Transaction Records (Raw Logs):")
    for idx, tx in enumerate(raw_transactions, 1):
        print(f"   [{idx:02d}] Customer: {tx['customer']} | Product: {tx['product']:<10} | Qty: {tx['quantity']}")

    print("\n2. Organized Transaction Data View:")
    for tx in raw_transactions:
        print(f"   Customer {tx['customer']} bought {tx['quantity']}x {tx['product']} (Unit Price: ${tx['unit_price']})")

    print("\n3. Repeated/Distinct Transactions:")
    print(f"   - Total Logged Transactions: {len(raw_transactions)}")
    print(f"   - Unique Customer-Product Relationships: {distinct_tx_count}")
    print("   - Duplicate Purchase Logs Detected:")
    for r_tx in repeated_txs:
        print(f"     * Customer {r_tx['customer']} re-ordered product: {r_tx['product']}")

    print("\n4. Individual Transaction Values:")
    for idx, tx in enumerate(raw_transactions, 1):
        print(f"   Log #{idx:02d}: {tx['quantity']} * ${tx['unit_price']:.2f} = Total Value: ${tx['total_value']:.2f}")

    print("\n5. Customer Summaries:")
    for cust, summary in customer_summaries.items():
        print(f"   - {cust:<12} | Total Items Purchased: {summary['total_items']:<3} | Total Revenue Spent: ${summary['total_spent']:.2f}")

    print("\n6. Product Summaries:")
    for prod, summary in product_summaries.items():
        print(f"   - {prod:<12} | Total Units Sold: {summary['units_sold']:<3} | Total Revenue Earned: ${summary['revenue']:.2f}")

    print("\n7. Ordered Results (Sorted Alphabetically by Customer & Product):")
    print(f"   {'Customer':<15} | {'Product':<12} | {'Quantity':<8} | {'Total Value'}")
    print("-" * 55)
    for tx in ordered_results:
        print(f"   {tx['customer']:<15} | {tx['product']:<12} | {tx['quantity']:<8} | ${tx['total_value']:.2f}")

    print("\n8. Overall Transaction Summary Dashboard:")
    overall_value = round(sum(t["total_value"] for t in raw_transactions), 2)
    overall_items = sum(t["quantity"] for t in raw_transactions)
    print(f"   - Grand Total Store Revenue Generated : ${overall_value:.2f}")
    print(f"   - Total Physical Units Distributed    : {overall_items} items")
    print("Final Output: Final Consolidated Transaction Report Compiled.")

if __name__ == "__main__":
    transaction_history_analyzer()
```
